### A.2.1 标量、向量、矩阵和张量
#### 代码清单 A-1 创建PyTorch张量

```python
import torch
tensor0d = torch.tensor(1) # 从 Python 整数创建一个零维张量（标量）
print(tensor0d)
```

```python
tensor1d = torch.tensor([1, 2, 3]) # 从 Python 列表创建一个一维张量（向量）
print(tensor1d)
```

```python
tensor2d = torch.tensor([[1, 2], [3, 4]]) # 从 Python 嵌套列表创建一个二维张量（矩阵）
print(tensor2d)
```

```python
tensor3d = torch.tensor([[[1, 2], [3, 4]], [[5, 6], [7, 8]]]) # 从 Python 嵌套列表创建一个三维张量
print(tensor3d)
```

### A.2.2 张量数据类型

```python
tensor1d = torch.tensor([1, 2, 3])
print(tensor1d.dtype)
```

```python
floatvec = torch.tensor([1.0, 2.0, 3.0])
print(floatvec.dtype)
```

### A.2.3 常见的 Pytorch 张量操作

```python
tensor2d = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(tensor2d)
# .shape 属性允许我们访问张量的形状
print(tensor2d.shape)
```

```python
# 使用 .reshape() 方法改变张量的形状
print(tensor2d.reshape(3, 2))
```

```python
# 在 PyTorch 中，重塑张量更常用的命令是 .view()
print(tensor2d.view(3, 2))
# .view() 方法与 .reshape() 方法不同，.view() 方法会改变张量的内存布局，而 .reshape() 方法会返回一个新的张量，可能会复制数据。因此，在使用 .view() 方法时，必须确保张量是连续的（即在内存中是连续存储的）。如果张量不是连续的，则需要先调用 .contiguous() 方法将其转换为连续的张量，然后再调用 .view() 方法。
```

```python
# 使用 .T 属性转置张量，沿对角线翻转
print(tensor2d.T)
```

```python
# PyTorch 中常用的矩阵相乘方法
# 使用 .matmul() 方法计算张量的乘积
print(tensor2d.matmul(tensor2d.T))
```

```python
# 也可以使用 @ 运算符计算张量的乘积
print(tensor2d @ tensor2d.T)
```
