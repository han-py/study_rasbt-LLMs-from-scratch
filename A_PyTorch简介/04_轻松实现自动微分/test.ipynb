### 练习1

```python
import torch

x = torch.tensor(3.0, requires_grad= True)
y = x ** 2

y.backward()

print(y)
print(x.grad) # 6
```

### 练习2

```python
x = torch.tensor(2.0, requires_grad=True)
y = x * 4
z = y + 1
loss = z ** 2 # (x * 4 + 1) ** 2

loss.backward()

print(loss)
print(x.grad) # 2 * (x * 4 +1) * 4 =72
```

### 练习3

```python
x = torch.tensor(1.0, requires_grad=True)
y1 = x * 2
y1.backward()
print(x.grad)

y2 = x * 3
y2.backward()
print(x.grad)
```

### 练习4

```python
x = torch.tensor(2.0, requires_grad=True)
y = x * 3
z = y.detach() * 5

print(z.requires_grad)
```
