### A.3 将模型视为计算图
#### 代码清单 A-2 逻辑回归的前向传播

```python
import torch
import torch.nn.functional as F # 在 PyTorch 中，这个导入语句是一种常见的约定，用于避免代码行过长

y = torch.tensor([1.0]) # 真实标签
x1 = torch.tensor([1.1]) # 输入特征
w1 = torch.tensor([2.2]) # 权重参数
b = torch.tensor([0.0]) # 偏置单元
z = x1 * w1 + b # 网络输入
a = torch.sigmoid(z) # 激活和输入
loss = F.binary_cross_entropy(a, y) # 二元交叉熵损失函数
print(loss) # 输出损失值
```
