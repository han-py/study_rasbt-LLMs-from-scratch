### A.6 设置高效的数据加载器
#### 代码清单 A-5 创建一个小型示例数据集

```python
import torch

x_train = torch.tensor([
    [-1.2, 3.1],
    [-0.9, 2.9],
    [-0.5, 2.6],
    [2.3, -1.1],
    [2.7, -1.5],
])
y_train = torch.tensor([0, 0, 0, 1, 1])

x_test = torch.tensor([
    [-0.8, 2.8],
    [2.6, -1.6],
])
y_test = torch.tensor([0, 1])
```

#### 代码清单 A-6 定义一个自定义的 Dataset 类

```python
from torch.utils.data import Dataset

class ToyDataset(Dataset):
    def __init__(self, x, y):
        self.features = x
        self.labels = y

    # 检索一条数据记录及其对应标签的说明
    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0] # 返回数据集总长度的说明

train_ids = ToyDataset(x_train, y_train)
test_ids = ToyDataset(x_test, y_test)
```

#### 代码清单 A-7 实例化数据加载器

```python
from torch.utils.data import DataLoader

torch.manual_seed(123)

train_loader = DataLoader(
    dataset=train_ids, # 之前创建的示例数据集实例作为数据加载器的输入
    batch_size=2,
    shuffle=True, # 是否打乱数据
    num_workers=0, # 后台进程的数量
)

test_loader = DataLoader(
    dataset=test_ids,
    batch_size=2,
    shuffle=False, # 测试数据不需要打乱
    num_workers=0,
)
```

```python
for idx, (x, y) in enumerate(train_loader):
    print(f'Batch {idx+1}:', x, y)
```

#### 代码清单 A-8 一个丢弃最后一个批次的数据加载器

```python
train_loader = DataLoader(
    dataset=train_ids,
    batch_size=2,
    shuffle=True,
    num_workers=0,
    drop_last=True, # 丢弃最后一个批次的数据
)
```

```python
for idx, (x, y) in enumerate(train_loader):
    print(f'Batch {idx+1}:', x, y)
```
